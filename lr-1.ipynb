## Часть 1 - Two Sum

Часть 1. Two Sum двумя способами — обязательная Решить Two Sum (https://leetcode.com/problems/two-sum/) на любом языке двумя способами:

a. «В лоб» — два вложенных цикла, отправить на LeetCode (Accepted), записать сложность O(n²).

b. Через HashMap — проходя, класть в карту значение → индекс, для текущего числа искать цель − число (в Python — обычный dict).

Записать вывод-сравнение (2–3 строки): почему HashMap выигрывает (O(n) против O(n²)) и что показало время прогона LeetCode.

### a. «В лоб»: два вложенных цикла

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        for i in range(n):
            for j in range(i + 1, n): 
                if nums[i] + nums[j] == target:
                    return [i, j]
        return []
```

Оценка: время O(n**2), память O(1)

Результат выполнения на leetcode: 

Runtime: 1780 ms, Beats: 10.21%

Memory: 19.47 MB, Beats: 99.90%

### b. HashMap

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {} 
        for i, x in enumerate(nums):
            need = target - x
            if need in seen: 
                return [seen[need], i]
            seen[x] = i
        return []
```

Оценка: Время O(n), память O(n)

Результат выполнения на leetcode: 

Runtime: 1 ms, Beats: 61.02%

Memory: 20.11 MB, Beats: 65.79%

### Причина разницы во времени выполнения

При решении задачи «в лоб» для каждого числа происходит перебор всех остальных, при n = 10⁴ это до n(n−1)/2 около 5·10⁷ проверок пар, из-за этого время выполнения 1780 мс. HashMap проходит массив один раз и на каждом шаге делает один поиск в словаре за O(1) в среднем: это ~10⁴ операций и 1 мс. 

Таким образом разница растёт вместе с n: при удвоении n перебор замедляется в 4 раза, HashMap в 2, то есть O(n²) против O(n).
