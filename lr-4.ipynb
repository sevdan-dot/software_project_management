# SQL выборки по StPlatform

### (a) все сдачи студента по фамилии (JOIN);

```sql
SELECT u.id, u.full_name, u.group_name, a.title, s.created_at, s.status
FROM submissions s
JOIN users u      
    ON u.id = s.student_id
JOIN assignment a 
    ON a.id = s.assignment_id
WHERE u.full_name LIKE 'Иванов %';
```

**Что делает:** 

Объединяем 3 таблицы INNER-ом по id студента и id задания, выполняем фильтрацию по ФИО начинающегося с Иванов. В выгрузку добавил `u.id`, чтобы убрать коллизии с однофамильцами в группе.

### (b) сколько сдач пришло на каждое задание (JOIN + COUNT + GROUP BY);

```sql
SELECT a.pair_number, a.title, COUNT(s.id)
FROM assignment a
LEFT JOIN submissions s 
    ON s.assignment_id = a.id
GROUP BY a.id, a.pair_number, a.title;
```

**Что делает:** 

Соединяем LEFT-ом 2 таблицы по id задания, группируем по `a.id, a.pair_number, a.title`, чтобы в SELECT можно было вывести ` a.pair_number, a.title` + использовать агригирующую функцию COUNT() над сдачами. В результате выходит, что SELECT выводит для каждого задания число его сдач, включая задания, на которые никто не сдал (у них 0).

### (c) студенты, не сдавшие ни одной работы (LEFT JOIN … IS NULL или NOT EXISTS, + роль) — вопрос недели «кого пинать».

```sql
SELECT u.id, u.full_name, u.group_name, u.status
FROM users u
LEFT JOIN submissions s ON s.student_id = u.id
WHERE u.role = 'STUDENT'
  AND s.id IS NULL
ORDER BY u.group_name, u.full_name;
```

**Что делает:** 

Присоединяем к каждому студенту его сдачи и оставляем тех, кому присоединить нечего (`s.id IS NULL`), то есть студентов без единой сдачи, с сортировкой по группе и ФИО. Также нужно добавить фильтр по роли STUDENT, т.к. возникнет забавная коллизия, что нужно пинговать преподавателя, чтобы он сдал работы.
